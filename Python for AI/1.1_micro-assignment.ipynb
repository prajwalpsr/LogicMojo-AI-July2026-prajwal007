Problem 1: Names report
1. Ada
2. Alan
3. Grace (long)
4. Kay
5. Katherine (long)
6. Ed

2 of 6 names are longer than four letters.

```python
names = ["Ada", "Alan", "Grace", "Kay", "Katherine", "Ed"]
# TODO: number each name; add " (long)" if len(name) > 4;
# then print "<n> of <total> names are longer than four letters."

total = len(names)
c = 0

for i, name in enumerate(names, start=1):
    if len(name)>4:
        print(f"{i}. {name + "(long)"}")
        c+=1
    else:
        print(f"{i}. {name}")

print(f"{c} of {total} names are longer than four letter")
```

Problem 2: Reading signs
12: positive
-5: negative
0: zero
8: positive
-3: negative
7: positive

positives: 3
total: 19

```python
readings = [12, -5, 0, 8, -3, 7]
positive = 0
sum_n = 0
# TODO: print "value: sign" (positive/negative/zero) for each;
# then print "positives: <count>" and "total: <sum>".

# your code here

for number in readings:
    sum_n += number
    if number>0:
        print(f"{number}: Positive")
        positive += 1
    elif number<0:
        print(f"{number}: negative")
    else:
        print(f"{number}: zero")

print(f"positives: {positive}")
print(f"total: {sum_n}")
```

Problem 4: Highest reading, without max()
highest: 47
first seen at index: 1
appears 3 times

```python
scores = [12, 47, 47, 8, 5, 47, 19]

# TODO (no max()): find the highest value, the index where it first appears,
# and how many times it appears. Track the "highest so far" as you scan,
# then count appearances in a second pass.

# your code here
highest = scores[0]
h_index = 0

for i, num in enumerate(scores):
    if highest<num:
        highest=num
        h_index=i

count = 0
for num in scores:
    if num == highest:
        count += 1


print(f"highest: {highest}\nfirst seen at index: {h_index} \nappears {count} times")
```
