```python
# TODO: define a class Account with __init__(self, balance=0) storing self.balance
# and a deposit(self, amount) method. Create two accounts (start 100 and 0), deposit
# 50 and 20, and print both balances to show independent state.

# your code here
class Account:
    def __init__(self, balance=0):
        self.balance = balance

    def deposite(self, amount):
        self.balance += amount
        return self.balance
```

```python
A = Account(100)
B = Account(0)

print(A.deposite(50))
print(B.deposite(20))
```

Problem 2: Sum the valid numbers
total: 20
skipped: 1

```python
raw = ["10", "7", "x", "3"]

# TODO: add up the items that are valid integers. Use try/except to catch the
# ValueError from int() on a bad item and skip it. Print the total and how many
# were skipped.

# your code here
total = 0
skipped = 0

for x in raw:
    try:
        total+=int(x)
    except ValueError:
        skipped+=1

print("Total:", total)
print("Skipped:", skipped)

```

Problem 4: A savings subclass
Account: 100
SavingsAccount: 200 at 0.05

```python
# TODO:
# 1. Define Account with __init__(self, balance) and summary() returning "Account: <balance>".
# 2. Define SavingsAccount(Account) that also takes a rate (call super().__init__ for the
#    balance) and overrides summary() to return "SavingsAccount: <balance> at <rate>".
# 3. Put one of each in a list and print each summary() (same call, different output).

# your code here

class Account:
    def __init__(self, balance):
        self.balance = balance

    def summary(self):
        return f"Account: {self.balance}"


class SavingsAccount(Account):
    def __init__(self, balance, rate):
        super().__init__(balance)
        self.rate = rate

    def summary(self):
        return f"SavigsAccount: {self.balance} at {self.rate}"

accounts = [
    Account(100),
    SavingsAccount(200, "5%")
]

for account in accounts:
    print(account.summary())
    
```
