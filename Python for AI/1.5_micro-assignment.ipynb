Problem 2: Check the status code

```python
responses = [
    {"status_code": 200, "data": {"answer": 42}},
    {"status_code": 404, "data": None},
]

# TODO: for each response, if status_code is 200 print "ok: ",
# otherwise print "request failed with status ".

# your code here
for res in responses:
    if res.get('status_code') == 200:
        print(f'ok: {res.get('data', "none")}')
    else:
        print(f"request failed with status: {res.get("status_code", "None")}")
```

Problem 4: Call a hosted model with Groq (run locally)

```python
import os
import requests
from dotenv import load_dotenv

load_dotenv()

# TODO: call Groq with the same prompt as Problem 3.
# Read the key from os.environ["GROQ_API_KEY"] (set it in your shell; never hardcode it).
# POST to https://api.groq.com/openai/v1/chat/completions with an Authorization header
# and a "messages" body, then read resp.json()["choices"][0]["message"]["content"].

# your code here
url = "https://api.groq.com/openai/v1/chat/completions"
api_key = os.environ["GROQ_API_KEY"]

headers = {
    "Authorization": f"Bearer {api_key}",
    "Content-Type": "application/json"
}


body = {
    "model": "openai/gpt-oss-120b",
    "messages": [
        {'role': 'system', 'content': 'You are a helpful assistant'},
        {'role': 'user', 'content': "Hello"}
    ]
}   

resp = requests.post(url, headers=headers, json=body).json()

print(resp.get('choices')[0].get("message")['content'])
```
