### Problem 2: Three-way split and model selection

```python
# Load load_wine. Split off 20% test (random_state=0, stratify=y), then split the rest
# into train/validation (test_size=0.25, random_state=0, stratify). Scale on train only.
# Train LogisticRegression(max_iter=5000) and DecisionTreeClassifier(max_depth=2, random_state=0),
# pick the higher validation accuracy, and print sizes, both validation accuracies, chosen, test accuracy.
# your code here

from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier

dataset = load_wine()
target_variable = dataset.target
dependent_variable = dataset.data

X_train, X_test, y_train, y_test = train_test_split(dependent_variable, target_variable, test_size=0.2, train_size=0.8, random_state=0, stratify=target_variable)

X_train, X_val, y_train, y_val = train_test_split(
    X_train,
    y_train,
    test_size=0.25,
    random_state=0,
    stratify=y_train
)

scalar = StandardScaler()

X_train = scalar.fit_transform(X_train)
X_test = scalar.transform(X_test)
X_val = scalar.transform(X_val)

linear_model = LogisticRegression(max_iter=5000)
decision_tree = DecisionTreeClassifier(max_depth=2, random_state=0)

linear_model.fit(X_train, y_train)
decision_tree.fit(X_train, y_train)

linear_acc = accuracy_score(
    y_val,
    linear_model.predict(X_val)
)

tree_acc = accuracy_score(
    y_val,
    decision_tree.predict(X_val)
)

if linear_acc > tree_acc:
    chosen_model = linear_model
    chosen = "logreg"
else:
    chosen_model = decision_tree
    chosen = "tree"

# Test accuracy
test_acc = accuracy_score(
    y_test,
    chosen_model.predict(X_test)
)

print(f"sizes train/val/test: {len(X_train)} {len(X_val)} {len(X_test)}")
print(f"validation logreg: {linear_acc:.2f}")
print(f"validation tree: {tree_acc:.2f}")
print(f"chosen: {chosen}")
print(f"test accuracy: {test_acc:.2f}")
```

### Problem 6: Precision and recall on imbalanced labels

```python
labels = [0] * 18 + [1, 1]
preds  = [0] * 20

TP = TN = FP = FN = 0

for label, pred in zip(labels, preds):
    if label == 1 and pred == 1:
        TP += 1
    elif label == 0 and pred == 0:
        TN += 1
    elif label == 0 and pred == 1:
        FP += 1
    elif label == 1 and pred == 0:
        FN += 1

accuracy = (TP + TN) / len(labels)

precision = TP / (TP + FP) if (TP + FP) > 0 else 0.0

recall = TP / (TP + FN) if (TP + FN) > 0 else 0.0

print(f"accuracy: {accuracy}")
print(f"precision: {precision}")
print(f"recall: {recall}")

print("accuracy flatters this model because the dataset is imbalanced and it predicts the majority class (0) for every sample.")
```
