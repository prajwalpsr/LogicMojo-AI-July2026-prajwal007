### Problem 4: The five-line loop lowers the loss

```python
import torch

class LinearRegression(torch.nn.Module):
    def __init__(self):
        super(LinearRegression, self).__init__()
        self.w = torch.nn.Parameter(torch.tensor(0.0, dtype=torch.float32))

    def forward(self, x):
        return self.w * x

x = torch.tensor([1., 2., 3., 4.])
y = torch.tensor([2., 4., 6., 8.])
w = torch.tensor([0.0], requires_grad=True)
# Fit w so w*x matches y: MSE loss, SGD(lr=0.02), 80 steps (the five-line loop).
# Print the first-step loss, the final loss, and whether it decreased.
# your code here
model = LinearRegression()

criterion = torch.nn.MSELoss()

optimizer = torch.optim.SGD(model.parameters(), lr=0.02)

initial_loss = criterion(model(x), y)
print(f"first loss: {initial_loss.item()}")

for step in range(80):
    prediction = model(x)

    loss = criterion(prediction, y)

    optimizer.zero_grad()

    loss.backward()

    optimizer.step()

final_loss = criterion(model(x), y)
print(f"last loss: {round(final_loss.item(), 2)}")

decreased = initial_loss > final_loss
print(f"decreased: {decreased}")
```

### Problem 5: The zero_grad trap

```python
# Start w = torch.tensor([1.0], requires_grad=True). Call (w**2).sum().backward() three
# times without resetting and record w.grad each time; then repeat resetting with w.grad.zero_().
# Print both lists.
# your code here

w = torch.tensor([1.0], requires_grad=True)
grads = []
wo_grad = []
for i in range(3):
    (w**2).sum().backward()
    grads.append(w.grad.item())

for i in range(3):
    (w**2).sum().backward()
    wo_grad.append(w.grad.item())
    w.grad.zero_()

print(f"without zero_grad: {grads},\nzero_grad: {wo_grad}")
```
